```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
```

```python
df = pd.read_csv("Loan_data.csv")
```

```python
print("Dataset Loaded Successfully!")
print(df)
```

```python
print("\nFirst 5 Rows")
print(df.head())
```

```python
print("\nDataset Information")
df.info()
```

```python
print("\nDataset Shape")
print(df.shape)
```

```python
print("\nStatistical Summary")
print(df.describe())
```

```python
print("\nMissing Values")
print(df.isnull().sum())
```

```python
df = df.fillna(method='ffill')
df = df.dropna()
print("\nMissing Values After Cleaning")
print(df.isnull().sum())
```

```python
print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("\nDataset After Removing Duplicates")
print(df)
```

```python
print("\nData Types")
print(df.dtypes)
```

```python
print("\nColumn Names")
print(df.columns)
```

```python
column = "LoanAmount"
median = df[column].median()

MAD = np.median(np.abs(df[column] - median))

print("Median =", median)
print("MAD =", MAD)


```

```python
modified_z = 0.6745 * (df[column] - median) / MAD

outliers = df[np.abs(modified_z) > 3.5]

print("\nDetected Outliers")
print(outliers)
print("\nNumber of Outliers:", len(outliers))
```

```python
df_cleaned = df[np.abs(modified_z) <= 3.5]

plt.figure(figsize=(6,4))
plt.boxplot(df_cleaned[column])
plt.title("After Outlier Removal")
plt.ylabel(column)
plt.show()
```

```python
print("\nOriginal Dataset Size:", df.shape)
print("Cleaned Dataset Size:", df_cleaned.shape)
print("Number of Outliers Removed:", len(df) - len(df_cleaned))
```

```python
df_cleaned.to_csv("cleaned_loan_data.csv", index=False)

print("\nCleaned Dataset Saved Successfully!")
```
